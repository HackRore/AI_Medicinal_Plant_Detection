## 🛠️ Phase 1: Universal Refinery (Normalization)
We consolidate species names from multiple datasets into a single dictionary mapping.

```python
def normalize_name(name):
    """Converts raw folder names to clean species names."""
    name = name.lower().replace("_", " ").replace("-", " ")
    name = name.replace("  ", " ").strip()
    return name.title()

def build_unified_repo():
    if not UNIFIED_DIR.exists(): UNIFIED_DIR.mkdir(parents=True)
    
    logger_dict = {}
    
    for ds_path in RAW_DIR.iterdir():
        if not ds_path.is_dir(): continue
        print(f"Processing {ds_path.name}...")
        
        # Handle nested folder variants (Dataset/ etc)
        search_root = ds_path
        if (ds_path / "Dataset").exists(): search_root = ds_path / "Dataset"
        
        for species_dir in search_root.iterdir():
            if not species_dir.is_dir(): continue
            
            clean_name = normalize_name(species_dir.name)
            target_dir = UNIFIED_DIR / clean_name
            target_dir.mkdir(exist_ok=True)
            
            # Mirror files (Sample first 500 per species to keep training balanced)
            files = list(species_dir.glob("*"))[:500]
            for f in files:
                shutil.copy(f, target_dir / f.name)
            
            logger_dict[clean_name] = logger_dict.get(clean_name, 0) + len(files)
    
    print("\nUnified Stats:")
    for k, v in sorted(logger_dict.items()):
        print(f"- {k}: {v} images")

# build_unified_repo()
```

## 🧠 Phase 2: Neural Architecture (EfficientNet-V2)
Transitioning to the Triple-Intelligence v3 core.

```python
class TripleIntelligenceV3(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.model = models.efficientnet_v2_s(weights='DEFAULT')
        
        # Fine-tune Head
        n_inputs = self.model.classifier[1].in_features
        self.model.classifier[1] = nn.Sequential(
            nn.Linear(n_inputs, 1024),
            nn.BatchNorm1d(1024),
            nn.ELU(),
            nn.Dropout(0.4),
            nn.Linear(1024, 512),
            nn.ELU(),
            nn.Linear(512, num_classes)
        )
        
    def forward(self, x):
        return self.model(x)

print("Model architecture defined. Optimized for Triple Dataset Ensemble.")
```
